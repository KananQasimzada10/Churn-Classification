```python

#They are not important for Machine learning model
dataset.drop(columns=['id', 'TelefonXidməti,KağızsızHesablaşma'], inplace=True)
```

```python
dataset['Ümumi ödənişlər'] = pd.to_numeric(dataset['Ümumi ödənişlər'], errors='coerce')
```

```python
data1 = dataset.copy()

from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy='most_frequent')
data1['cinsiyyət'] = imputer.fit_transform(data1[['cinsiyyət']]).ravel()
data1['tərəfdaş'] = imputer.fit_transform(data1[['tərəfdaş']]).ravel()
```

```python
data1.isnull().sum()
```

```python
data1 = data1.dropna(subset=['Ümumi ödənişlər'])
```

```python
data1.isnull().sum()
```

```python
#Check outliers in 'müddəti' column(Result:Empty)
#Method with boxplot
import matplotlib.pyplot as plt
plt.boxplot(data1['müddəti'])
plt.show()
```

```python
Q1 = data1['Aylıq Ödənişlər'].quantile(0.25)
Q3 = data1['Aylıq Ödənişlər'].quantile(0.75)
IQR = Q3 - Q1

#Define outliers for 'Aylıq Ödənişlər'
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers1 = data1[(data1['Aylıq Ödənişlər'] < lower_bound) | (data1['Aylıq Ödənişlər'] > upper_bound)]

print(outliers1)
```

```python
from scipy.stats import zscore

data1 = data1.dropna(subset=['Ümumi ödənişlər'])

# Calculate the z-scores for 'Ümumi ödənişlər' and add it to data1
data1['z_score'] = zscore(data1['Ümumi ödənişlər'])

# Identify outliers based on the z-score in data1 (the corrected part)
outliers2 = data1[abs(data1['z_score']) > 3]

print(outliers2)

# Drop z_score column, since the question asks for this
data1.drop(columns=['z_score'], inplace=True)
```

```python
#Encode process for columns

data2 = data1.copy()

from sklearn.preprocessing import LabelEncoder

categorical_columns = data2.select_dtypes(include=['object']).columns
binary_columns = categorical_columns[(data2[categorical_columns].nunique() == 2) & (categorical_columns != 'Churn')]

print(binary_columns)

label_encoders = {}

for column in binary_columns:
  le = LabelEncoder()
  data2[column] = le.fit_transform(data2[column])
  label_encoders[column] = le  # Store the encoder for future reference

```

```python
data2.head()
```

```python
import pandas as pd

# Step 1: Select categorical columns with more than 2 unique values, excluding 'churn'
categorical_columns = data2.select_dtypes(include=['object']).columns
non_binary_columns = categorical_columns[(data2[categorical_columns].nunique() > 2) ]

# Step 2: Apply OneHot Encoding
data2 = pd.get_dummies(data2, columns=non_binary_columns, drop_first=False)

print("OneHot encoding applied using pd.get_dummies.")
```

```python
data2.head()
```

```python
data2['Churn'] = data2['Churn'].replace({'Hə': 1, 'Yox': 0})
```

```python
data2.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

# Select columns to normalize
columns_to_normalize = ['müddəti', 'Aylıq Ödənişlər', 'Ümumi ödənişlər']

# Apply Min-Max Scaling
scaler = MinMaxScaler()
data2[columns_to_normalize] = scaler.fit_transform(data2[columns_to_normalize])

# Check the normalized dataset
print(data2[columns_to_normalize].head())
```

```python
data2.head()
```

```python
pd.DataFrame(data2).to_excel("dataset_preprocessing.xlsx", index=False)
```

```python
from sklearn.model_selection import train_test_split
X = data2.drop('Churn', axis=1)
y = data2['Churn']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)
```

```python
# Count the instances in each class after SMOTE
class_counts_resampled = y_train_resampled.value_counts()
print("Class Counts After SMOTE:\n", class_counts_resampled)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(X_train_resampled, y_train_resampled)


# Random Forest Metrics
y_pred_rf = rf_model.predict(X_train_resampled)
y_pred_rf_prob = rf_model.predict_proba(X_train_resampled)[:, 1]
rf_accuracy = accuracy_score(y_train_resampled, y_pred_rf)
rf_precision = precision_score(y_train_resampled, y_pred_rf)
rf_recall = recall_score(y_train_resampled, y_pred_rf)
rf_f1 = f1_score(y_train_resampled, y_pred_rf)
rf_roc_auc = roc_auc_score(y_train_resampled, y_pred_rf_prob)




```

```python
# Print metrics
print("Random Forest Metrics:")
print(f"Accuracy: {rf_accuracy:}")
print(f"Precision: {rf_precision:}")
print(f"Recall: {rf_recall:}")
print(f"F1 Score: {rf_f1:}")
print(f"ROC-AUC: {rf_roc_auc:}")
```

```python
from sklearn.svm import SVC

svm_model = SVC(probability=True, random_state=42)
svm_model.fit(X_train_resampled, y_train_resampled)


# SVM Metrics
y_pred_svm = svm_model.predict(X_train_resampled)
y_pred_svm_prob = svm_model.predict_proba(X_train_resampled)[:, 1]  # SVM with probability=True
svm_accuracy = accuracy_score(y_train_resampled, y_pred_svm)
svm_precision = precision_score(y_train_resampled, y_pred_svm)
svm_recall = recall_score(y_train_resampled, y_pred_svm)
svm_f1 = f1_score(y_train_resampled, y_pred_svm)
svm_roc_auc = roc_auc_score(y_train_resampled, y_pred_svm_prob)
```

```python
# Print metrics
print("SVM Metrics:")
print(f"Accuracy: {svm_accuracy:}")
print(f"Precision: {svm_precision:}")
print(f"Recall: {svm_recall:}")
print(f"F1 Score: {svm_f1:}")
print(f"ROC-AUC: {svm_roc_auc:}")
```

```python
from xgboost import XGBClassifier
xgb_model = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss')
xgb_model.fit(X_train_resampled, y_train_resampled)


# XGBoost Metrics
y_pred_xgb = xgb_model.predict(X_train_resampled)
y_pred_xgb_prob = xgb_model.predict_proba(X_train_resampled)[:, 1]
xgb_accuracy = accuracy_score(y_train_resampled, y_pred_xgb)
xgb_precision = precision_score(y_train_resampled, y_pred_xgb)
xgb_recall = recall_score(y_train_resampled, y_pred_xgb)
xgb_f1 = f1_score(y_train_resampled, y_pred_xgb)
xgb_roc_auc = roc_auc_score(y_train_resampled, y_pred_xgb_prob)
```

```python
# Print metrics
print("SVM Metrics:")
print(f"Accuracy: {xgb_accuracy:}")
print(f"Precision: {xgb_precision:}")
print(f"Recall: {xgb_recall:}")
print(f"F1 Score: {xgb_f1:}")
print(f"ROC-AUC: {xgb_roc_auc:}")
```

```python
from sklearn.linear_model import LogisticRegression

# Logistic Regression Classifier
lr_model = LogisticRegression(random_state=42, max_iter=1000)
lr_model.fit(X_train_resampled, y_train_resampled)


# Logistic Regression Metrics
y_pred_lr = lr_model.predict(X_train_resampled)
y_pred_lr_prob = lr_model.predict_proba(X_train_resampled)[:, 1]
lr_accuracy = accuracy_score(y_train_resampled, y_pred_lr)
lr_precision = precision_score(y_train_resampled, y_pred_lr)
lr_recall = recall_score(y_train_resampled, y_pred_lr)
lr_f1 = f1_score(y_train_resampled, y_pred_lr)
lr_roc_auc = roc_auc_score(y_train_resampled, y_pred_lr_prob)
```

```python
# Import necessary libraries
from sklearn.model_selection import train_test_split
from lazypredict.Supervised import LazyClassifier
from imblearn.over_sampling import SMOTE
import pandas as pd

# Verify if the 'Churn' column exists in data2
assert 'Churn' in data2.columns, "Column 'Churn' not found in dataset!"

# Split dataset into features (X) and target (y)
X = data2.drop(columns=['Churn'])
y = data2['Churn']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Apply SMOTE to the training set
smote = SMOTE(random_state=42)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train, y_train)

# Initialize LazyClassifier
lazy_clf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)

# Fit LazyClassifier on the resampled training set
models, predictions = lazy_clf.fit(X_train_resampled, X_test, y_train_resampled, y_test)

# Display results using pandas
print("Lazy Prediction After SMOTE:")
print(models)

# Save results to a CSV file for future reference
models.to_csv('lazy_prediction_results.csv', index=False)
print("Results saved to 'lazy_prediction_results.csv'")
```
